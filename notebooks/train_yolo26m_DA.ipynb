```python
from ultralytics import YOLO

model = YOLO("yolo26m.pt")

model.train(
    data="CarDD_YOLO_AUG/dataset.yaml",
    name="yolo26m_aug",
    epochs=300,
    patience=50,
    imgsz=640,
    batch=16,
    seed=42,
    lr0=0.001,
    workers=8,
    cache=True,
)
```
